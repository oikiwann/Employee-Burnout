```python
import pandas as pd
import numpy as np
df = pd.read_csv("data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv")
```

```python
df.map(lambda x: x.strip() if isinstance(x, str) else x) # remove leading and trailing whitespace from string columns
df.dropna(inplace=True)  # remove rows with missing values
df.drop_duplicates(inplace=True) # remove duplicate rows
cols_to_convert = ['Age', 'DailyRate', 'DistanceFromHome', 'Education', 'EmployeeCount', 'EmployeeNumber', 'EnvironmentSatisfaction','HourlyRate', 'JobInvolvement', 'JobLevel', 'JobSatisfaction', 'MonthlyIncome', 'MonthlyRate', 'NumCompaniesWorked', 'PercentSalaryHike', 'PerformanceRating', 'RelationshipSatisfaction', 'StandardHours', 'StockOptionLevel', 'TotalWorkingYears', 'TrainingTimesLastYear', 'WorkLifeBalance', 'YearsAtCompany', 'YearsInCurrentRole', 'YearsSinceLastPromotion', 'YearsWithCurrManager']
df[cols_to_convert] = df[cols_to_convert].astype(int) # convert columns to int type
```

```python
df["Gender"] = df["Gender"].replace({'Male': 'M', 'Female': 'F'}) #make tenure groups for experience level
df["OverTime"] = df["OverTime"].map({"Yes" : 1, "No" : 0}) # Transform the OverTime column to binary values (1 for Yes, 0 for No)
```

```python
conditions = [df["YearsAtCompany"] < 3, (df["YearsAtCompany"] >= 3) & (df["YearsAtCompany"] < 7), df["YearsAtCompany"] >= 7]
choices = ['Junior', 'Mid-level', 'Senior']
df["ExperienceLevel"] = np.select(conditions, choices, default='Unknown') # make tenure groups for experience level
```

```python
# burnout rate calculation is 0-5 scoring for burnout rate based on the following colums: EnvironmentSatisfaction, JobInnvolvement, OverTime, JobSatisfaction, WorkLifeBalance, 
df["burnout_rate"] = (df["EnvironmentSatisfaction"] + df["JobInvolvement"] + df["OverTime"] + df["JobSatisfaction"] + df["WorkLifeBalance"])/5
```

```python
df["AnnualIncome"] = df["MonthlyIncome"] * 12 # calculate annual income from monthly income
```

```python
df.to_csv('cleaned_data.csv', index=False)
```
